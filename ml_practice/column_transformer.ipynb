```python
from sklearn.compose import ColumnTransformer
transformer = ColumnTransformer(transformers= [('tnf1', SimpleImputer(), ['fever']),('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),('tnf3', OneHotEncoder(drop='first', sparse_output=True),['gender', 'city'])], remainder='passthrough')
```

```python
xtrain_transformed = transformer.fit_transform(xtrain)
```
