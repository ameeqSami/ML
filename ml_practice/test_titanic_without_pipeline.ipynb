```python
import pickle
ohe_sex = pickle.load(open('ohe_sex.pkl', 'rb'))
ohe_embarked = pickle.load(open('ohe_embarked.pkl', 'rb'))
clf = pickle.load(open('clf.pkl', 'rb'))
```

```python
import numpy as np
test_input = [2, 'male', 31.0, 0, 0, 10.5, 'S']
```

```python
test_input
```

```python
sex_encoded = ohe_sex.transform(np.array(test_input[1]).reshape(1, -1))
embarked_encoded = ohe_embarked.transform(np.array(test_input[6]).reshape(1, -1))

age = np.array(test_input[0]).reshape(1, 1)
sibsp = np.array(test_input[2]).reshape(1, 1)
parch = np.array(test_input[3]).reshape(1, 1)
fare = np.array(test_input[4]).reshape(1, 1)
class_val = np.array(test_input[5]).reshape(1, 1)

test_transformed = np.concatenate(
    (age, sex_encoded, sibsp, parch, fare, class_val, embarked_encoded), 
    axis=1
)

print(test_transformed.shape)
```

```python
clf.predict(test_transformed)
```
