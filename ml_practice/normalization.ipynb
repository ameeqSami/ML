```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
data = { 'weights' : [130, 67, 81, 32, 54]}
```

```python
df = pd.DataFrame(data)
```

```python
df.info()
```

```python
df['weights'] =  df['weights'].astype(float)
min_val = df['weights'].min()
max_val = df['weights'].max()

for i in range(0, len(df)):
    df.loc[i,'weights'] = ( df.loc[i,'weights'] - min_val)/(max_val - min_val)

for vals in df['weights']:
    print(vals)
```
